```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_excel('/content/covid_19_clean_complete.csv.xlsx')
```

```python
# Calculate total confirmed cases per country
country_confirmed = data.groupby('Country/Region')['Confirmed'].sum().sort_values(ascending=False)

# Select the top 10 countries
top_10_countries = country_confirmed.head(10)

plt.figure(figsize=(12, 6))
sns.barplot(y=top_10_countries.index, x=top_10_countries.values, palette='viridis')
plt.title('Top 10 Countries by Confirmed Cases')
plt.xlabel('Total Confirmed Cases')
plt.ylabel('Country/Region')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
